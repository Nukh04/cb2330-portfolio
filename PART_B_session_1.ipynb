```python
###Chosen formula: Running Total

import numpy as np
Example_Array = [0,4,8,25,3,17]
total = 0
for i in range (len(Example_Array)):
  total += Example_Array[i]
  print("Position:",i,"Value:",total)

Numpy_Array = np.array(Example_Array)
Numpy_Sum = np.sum(Numpy_Array)
if total == Numpy_Sum:
  print("They are the same value!")
```
